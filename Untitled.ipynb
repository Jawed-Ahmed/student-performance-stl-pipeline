```python
# transformation

student_df = df.copy()       #Don't modify the original DataFrame directly
```

```python
student_df = student_df.drop_duplicates()   #remove duplicate
student_df.shape
```

```python
student_df[
    (student_df["attendance"] < 0) |
    (student_df["attendance"] > 100)
]
```

```python
marks_columns = [
    "math",
    "science",
    "english",
    "programming",
    "assignment_score",
    "midterm_score",
    "final_exam_score"
]

for column in marks_columns:
    invalid = student_df[
        (student_df[column] < 0) |
        (student_df[column] > 100)
    ]

    if len(invalid) > 0:
        print(column)
        print(invalid)
```

```python
#Replace invalid attendance
#First calculate the valid average

valid_attendance_mean = student_df[
    (student_df["attendance"] >= 0) &
    (student_df["attendance"] <= 100)
]["attendance"].mean()

#then
student_df.loc[
    (student_df["attendance"] < 0) |
    (student_df["attendance"] > 100),
    "attendance"
] = valid_attendance_mean
```

```python
#Replace invalid Math
#Calculate the valid mean
valid_math_mean = student_df[
    (student_df["math"] >= 0) &
    (student_df["math"] <= 100)
]["math"].mean()

#then
 student_df.loc[
    (student_df["math"] < 0) |
    (student_df["math"] > 100),
    "math"
] = valid_math_mean              #Now the Math values should be valid.
```

```python
#HANDLE MISSING VALUES
student_df.isnull().sum()
#handle the missing numerical values
for column in marks_columns:
    student_df[column] = student_df[column].fillna(
        student_df[column].mean()
    )

#attendance
student_df["attendance"] = student_df["attendance"].fillna(
    student_df["attendance"].mean()
)
#check again
student_df.isnull().sum()
```

```python
#transform data  - clean data i.ie create new useful information
student_df["total_marks"] = (
    student_df["math"] +
    student_df["science"] +
    student_df["english"] +
    student_df["programming"]
)
student_df[
    [
        "student_id",
        "math",
        "science",
        "english",
        "programming",
        "total_marks"
    ]
].head()
```

```python
student_df["percentage"] = (
    student_df["total_marks"] / 400
) * 100
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"
student_df["grade"] = student_df["percentage"].apply(
    calculate_grade
)
student_df["result"] = student_df["percentage"].apply(lambda x: "Pass" if x >= 40 else "Fail")
def performance_category(percentage):
    if percentage >= 90:
        return "Excellent"
    elif percentage >= 75:
        return "Good"
    elif percentage >= 60:
        return "Average"
    else:
        return "Needs Improvement"
student_df["performance_category"] = student_df["percentage"].apply(performance_category)
```
